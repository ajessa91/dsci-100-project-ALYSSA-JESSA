Visualization 1: %%% EDIT THIS MARKDOWN CELL%%%

```r
ggplot(basketball_tidy, aes(x = seed, fill = first_game_win)) +
  geom_bar(position = "fill") +
  scale_fill_manual(
    values = c("FALSE" = "#FA8072", "TRUE" = "#75de75"),
    labels = c("FALSE" = "Lost", "TRUE" = "Won")
  ) +
  labs(
    title = "First Tournament Game Outcomes by Seed",
    x = "Tournament Seed",
    y = "Proportion of Teams",
    fill = "First Game Outcome"
  )
```

Interpretation: 
This graph shows the relationship between a teams tournament seed and whether they won their first game. Teams with lower number seeds have a higher proportion of wins, while teams with higher number seeds are more likely to lose. This suggests that tournament seed could be a useful predictor of whether a team wins its first game.


Visualization 2: %%% EDIT THIS MARKDOWN CELL%%%

```r
ggplot(basketball_tidy, aes(x = first_game_at_home, fill = first_game_win)) +
  geom_bar(position = "fill") +
  scale_fill_manual(
    values = c("FALSE" = "#FA8072", "TRUE" = "#75de75"),
    labels = c("FALSE" = "Lost", "TRUE" = "Won")
  ) +
  scale_x_discrete(
    labels = c("N" = "Away", "Y" = "Home")
  ) +
  labs(
    title = "First Tournament Game Outcomes by Home Court",
    x = "First Game Location",
    y = "Proportion of Teams",
    fill = "First Game Outcome"
  )
```

Interpretation: 
The second graph compares first game outcomes for teams that played at home versus teams that played away. Teams playing at home had a much higher proportion of wins compared to teams playing away. This suggests that playing at home may be associated with a greater chance of winning the first tournament game and could be helpful predictor.
